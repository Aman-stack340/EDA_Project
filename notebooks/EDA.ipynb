```python
highest = df[df["Year"] == 2025].nlargest(
    1, "Life evaluation (3-year average)"
)[["Country name", "Life evaluation (3-year average)"]]

lowest = df[df["Year"] == 2025].nsmallest(
    1, "Life evaluation (3-year average)"
)[["Country name", "Life evaluation (3-year average)"]]

highest_score = highest["Life evaluation (3-year average)"].iloc[0]
lowest_score = lowest["Life evaluation (3-year average)"].iloc[0]

gap = highest_score - lowest_score

print("Highest:")
print(highest.to_string(index=False))

print("\nLowest:")
print(lowest.to_string(index=False))

print("\nScore Gap:", round(gap, 3))
```

```python
comparison = pd.concat([highest, lowest])

plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Country name"],
    comparison["Life evaluation (3-year average)"]
)

plt.title("Highest vs Lowest Life Evaluation - 2025")
plt.xlabel("Country")
plt.ylabel("Life Evaluation Score")

plt.show()
```

# Exploratory Data Analysis: World Happiness Report 2026

## 1. Project Overview

This project performs Exploratory Data Analysis (EDA) on the World Happiness Report 2026 dataset. The dataset contains country-level life evaluation scores and several explanatory factors for the years 2011–2025.

The objective is to identify trends, distributions, relationships, and important factors associated with life evaluation.

---

## 2. Dataset Overview

- Number of records: 2,116
- Number of columns: 13
- Number of unique countries: 168
- Time period: 2011–2025
- Target variable: Life evaluation (3-year average)

The dataset contains information about:

- Life evaluation
- GDP per capita
- Social support
- Healthy life expectancy
- Freedom to make life choices
- Generosity
- Perceptions of corruption
- Dystopia + residual

---

## 3. Data Quality Analysis

The dataset contains no duplicate rows.

The main life-evaluation variable contains no missing values.

Several explanatory variables contain substantial missing values for earlier years. These missing values are systematic rather than random. Most explanatory variables are unavailable for 2011–2018, while they are largely available from 2019 onward.

Therefore, missing explanatory values were not artificially imputed.

Correlation and factor analysis was performed using data from 2019–2025.

---

## 4. Descriptive Analysis

The average global life-evaluation score was approximately 5.39 in 2011 and approximately 5.65 in 2025.

The yearly averages show some fluctuations, but the overall long-term pattern is upward.

The lowest yearly average in the available data occurred around 2016, while 2025 had the highest yearly average.

---

## 5. Trend Analysis

The average life-evaluation scores were:

- 2011: 5.392
- 2012: 5.419
- 2014: 5.376
- 2015: 5.382
- 2016: 5.354
- 2017: 5.376
- 2018: 5.407
- 2019: 5.473
- 2020: 5.533
- 2021: 5.554
- 2022: 5.540
- 2023: 5.528
- 2024: 5.578
- 2025: 5.650

Overall, the average life-evaluation score increased between 2011 and 2025, although the increase was not completely consistent every year.

---

## 6. Top Countries in 2025

The highest life-evaluation scores in 2025 were observed for:

1. Finland
2. Iceland
3. Denmark
4. Costa Rica
5. Sweden
6. Norway
7. Netherlands
8. Israel
9. Luxembourg
10. Switzerland

Finland had the highest score in the dataset for 2025, with a life-evaluation score of 7.764.

---

## 7. Correlation Analysis

Correlation analysis was performed using 2019–2025 data.

The strongest correlations with life evaluation among the explanatory factors were:

| Factor | Correlation |
|---|---:|
| Social support | 0.701 |
| Log GDP per capita | 0.681 |
| Healthy life expectancy | 0.659 |
| Freedom to make life choices | 0.491 |
| Perceptions of corruption | 0.426 |
| Dystopia + residual | 0.378 |
| Generosity | 0.035 |

Social support had the strongest positive correlation with life evaluation among the explanatory factors, followed by log GDP per capita and healthy life expectancy.

These correlations indicate statistical association and should not be interpreted as proof of causation.

---

## 8. Major Insights

### Insight 1: Long-term increase

Average life evaluation increased from approximately 5.39 in 2011 to approximately 5.65 in 2025.

### Insight 2: Social support

Social support had the strongest correlation with life evaluation among the explanatory factors, with a correlation of approximately 0.701.

### Insight 3: Economic conditions

Log GDP per capita showed a strong positive correlation of approximately 0.681 with life evaluation.

### Insight 4: Health

Healthy life expectancy also showed a strong positive correlation of approximately 0.659.

### Insight 5: Freedom

Freedom to make life choices showed a moderate positive correlation of approximately 0.491.

### Insight 6: Generosity

Generosity showed very little linear correlation with life evaluation in this dataset, with a correlation of approximately 0.035.

---

## 9. Conclusion

The exploratory analysis indicates that life evaluation varies considerably across countries and years.

The overall average life-evaluation score shows an upward pattern between 2011 and 2025. Analysis of the 2019–2025 data indicates that social support, log GDP per capita, and healthy life expectancy have the strongest positive associations with life evaluation among the explanatory factors examined.

The analysis also demonstrates the importance of understanding missing data before performing statistical analysis. Because explanatory variables were largely unavailable for earlier years, the correlation analysis was restricted to the period where these variables were available rather than filling the missing values artificially.

Overall, the EDA demonstrates how statistical summaries, visualizations, trend analysis, and correlation analysis can be combined to identify meaningful patterns in a real-world dataset.
